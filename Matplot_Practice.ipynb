```python
# import Matplot Lib
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(1,6)
y = np.array([2,4,6,8,10])
print(x)
print(y)
```

```python
plt.figure(figsize=(12,8)) # creates a new Matplotlib figure and sets its dimensions to 12 inches wide by 8 inches tall
```

```python
# Line Plot
plt.subplot(2,3,1)  # plt.subplot(nrows, ncols, index) -- Grid will have 2 rows , Grid will have 3 columns , Position of this specific plot, ( starting at 1 in the top-left and counting left-to-right, top-to-bottom )
plt.plot(x,y)
plt.title("Line Plot")
```

```python
# Scatter plot
plt.subplot(2,3,2)
plt.scatter(x,y)
plt.title("Scatter Plot")
```

```python
# Histogram
plt.subplot(2,3,3)
plt.hist(x,y)
plt.title("Histogram")
```

```python
# Bar plot
plt.subplot(2,3,4)
plt.bar(x,y)
plt.title("Bar Plot")
```

```python
# Box plot
plt.subplot(2,3,6)
plt.boxplot(y)
plt.title("Box Plot")
```

```python
# Pie Chart
plt.subplot(2,3,6)
plt.pie(y,labels=x,autopct='%1.1f%%')
plt.title("Pie Chart")
```

```python
plt.tight_layout()
plt.show()
```
