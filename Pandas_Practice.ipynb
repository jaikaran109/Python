```python
# Import Pandas
import pandas as pd
```

```python
# Create a Simple DataFrame
data = {
    'Name' : ['Chutki','Bholu','Kaliya','Motu','Bheem'],
    'Age' : [18,19,28,38,90],
    'Score' : [89,78,76,45,99]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Basic Operation
print(df.head(3))  # first 3 rows
print(df.tail(2)) # last 2 rows
print(df.info())  # info about DataFrame
print(df.describe())  # Summary statics
```

```python
# Value Counts
print(df['Age'].value_counts())
print(df['Name'].value_counts)
```

```python
# Drop Operations
df_drop = df.drop('Score',axis=1) # drop the Score -- axis 1 for Columns and axis 0 for Rows
print(df_drop)

df_dropna = df.dropna() # dropna() is a Pandas method used to remove missing or null values (NaN, None) from a DataFrame or Series.
print(df_dropna)

df_unique = df.drop_duplicates() # ye Duplicates remove kr dega
print(df_unique)
```

```python
# Rename Column
df_rename = df.rename(columns = {'Age':'Years'})  # Age ko Years name me change kr diya
print(df_rename)
```
